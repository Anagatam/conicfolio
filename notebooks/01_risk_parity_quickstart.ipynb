# ConicFolio Quickstart: Equal Risk Contribution (ERC) / Risk Parity

This tutorial demonstrates how to solve a true **Equal Risk Contribution (ERC)** portfolio using `conicfolio` in less than **1 millisecond** on 5 years of real institutional market returns.

### Why Risk Parity Matters
In a traditional 60/40 capital allocation, over **90% of total portfolio volatility risk** comes from equities. ConicFolio solves for allocations where every asset or sector contributes an **identical fraction of risk**, avoiding catastrophic equity drawdowns.

```python
# Install the public thin-client SDK
# !pip install conicfolio

import conicfolio as cf
import numpy as np
import pandas as pd
```

```python
# 1. Load real institutional multi-asset daily returns (1,257 trading days)
returns = cf.load_market_data(n_assets=10)
print(f"Loaded dataset: {returns.shape[0]} sessions across {returns.shape[1]} instruments")
returns.head()
```

```python
# 2. Initialize and fit the Risk Parity model
# Optional: cf.set_api_key('cf_live_YOUR_KEY')
model = cf.RiskParity()
model.fit(returns)

print("=== CONICFOLIO SOLVE REPORT ===")
print(f"Status: Certified Optimal")
print(f"Solve Time: {model.solve_time_ms_} ms")
print(f"Effective Number of Assets: {model.effective_assets_:.2f}")
print("Allocations:")
for asset, weight in zip(returns.columns, model.weights_):
    print(f"  {asset}: {weight * 100:.2f}%")
```
