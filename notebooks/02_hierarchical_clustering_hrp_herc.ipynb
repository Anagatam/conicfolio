# Hierarchical Risk Parity (HRP) & Hierarchical Equal Risk Contribution (HERC)

Modern quantitative portfolio optimization uses **graph and tree clustering** to eliminate the need for matrix inversion, preventing noise amplification in high-dimensional portfolios.

```python
import conicfolio as cf

returns = cf.load_market_data(n_assets=15)

# Fit Hierarchical Equal Risk Contribution with 3 clusters
herc = cf.HierarchicalEqualRiskContribution(n_clusters=3)
herc.fit(returns)

print(f"HERC solved in {herc.solve_time_ms_} ms")
print("Optimal Weights:", herc.weights_)
```
